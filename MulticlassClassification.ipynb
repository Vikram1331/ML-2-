```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["target"] = iris.target
df["flower"] = df["target"].map(
    {0: "setosa", 1: "versicolor", 2: "virginica"}
)

df.to_csv("iris.csv", index=False)

print(df)
```
